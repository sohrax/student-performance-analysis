# Our project is on creating a Student performance Analyser model---------

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {'hours_studied': [1,2,3,4,5,6,7,8,2,6],
    'sleep_hours': [7,6,6,5,5,6,7,8,6,5],
    'attendance': [60,65,70,75,80,85,90,95,70,88],
    'marks': [30,40,50,55,65,70,80,90,48,75]}
df = pd.DataFrame(data)
df
```

```python
plt.scatter(df['hours_studied'], df['marks'],color="red")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours students studied vs the marks they got")
plt.show()
```

# Now this is where things get more than just too basic...........it's still basic, but, something new to me

```python
#time for linear regression
from sklearn.linear_model import LinearRegression

X = df[['hours_studied', 'sleep_hours', 'attendance']]
y = df['marks']

model = LinearRegression()
model.fit(X, y)
```

```python
#Now we predict the marks of students
input_data = pd.DataFrame([[5,6,80]],columns=['hours_studied', 'sleep_hours', 'attendance'])
pred = model.predict(input_data)
print("Predicted Marks:", pred)
```

# Okay,so the model is working.....it's a simple one.....but we always start with simple isn't it?.....so lessgooooo....

```python
#What about we now let the user input values by themsleves?
hours = float(input("Enter study hours per day: "))
sleep = float(input("Enter sleep hours: "))
attendance = float(input("Enter attendance (%): "))
input1_data = pd.DataFrame([[hours, sleep, attendance]], 
                          columns=['hours_studied', 'sleep_hours', 'attendance'])
prediction = model.predict(input1_data)
print("Predicted Marks:", round(prediction[0], 2))
```
